# Train Baseline ANN (MNIST)

This notebook trains a baseline ANN on MNIST, saves the best model, and evaluates performance.


```python
import matplotlib.pyplot as plt

from src.data.data_loader import build_tf_datasets
from src.evaluation.metrics import evaluate_model
from src.models.ann import build_ann_model
from src.training.trainer import train_model
from src.utils.config import DEFAULT_DATA_CONFIG
```

```python
datasets = build_tf_datasets(DEFAULT_DATA_CONFIG)
model = build_ann_model(DEFAULT_DATA_CONFIG)
history, best_model_path = train_model(model, datasets)
print(f"Best model saved to: {best_model_path}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(history.history["loss"], label="train")
axes[0].plot(history.history["val_loss"], label="val")
axes[0].set_title("Loss")
axes[0].legend()

axes[1].plot(history.history["accuracy"], label="train")
axes[1].plot(history.history["val_accuracy"], label="val")
axes[1].set_title("Accuracy")
axes[1].legend()
plt.show()
```

```python
evaluation = evaluate_model(model, datasets["test"])
print("Test loss:", evaluation["loss"])
print("Test accuracy:", evaluation["accuracy"])
print("Confusion matrix:\n", evaluation["confusion_matrix"])
print("Classification report:\n", evaluation["classification_report"])
```
