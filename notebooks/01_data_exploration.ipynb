# Data Exploration (MNIST)

This notebook explores the MNIST dataset and validates the preprocessing pipeline.


```python
import matplotlib.pyplot as plt
import numpy as np

from src.data.data_loader import load_mnist
from src.utils.config import DEFAULT_DATA_CONFIG
```

```python
splits = load_mnist(DEFAULT_DATA_CONFIG)
(train_images, train_labels) = splits["train"]
(val_images, val_labels) = splits["validation"]
(test_images, test_labels) = splits["test"]

print("Train shapes:", train_images.shape, train_labels.shape)
print("Validation shapes:", val_images.shape, val_labels.shape)
print("Test shapes:", test_images.shape, test_labels.shape)
```

```python
fig, axes = plt.subplots(1, 5, figsize=(10, 2))
for idx, ax in enumerate(axes):
    ax.imshow(train_images[idx].squeeze(), cmap="gray")
    ax.set_title(f"Label: {np.argmax(train_labels[idx])}")
    ax.axis("off")
plt.show()
```

```python
class_counts = train_labels.sum(axis=0)
plt.bar(range(DEFAULT_DATA_CONFIG.num_classes), class_counts)
plt.title("Class Distribution (Train)")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
print("NaNs in train images:", np.isnan(train_images).any())
print("NaNs in train labels:", np.isnan(train_labels).any())
print("Pixel range:", train_images.min(), train_images.max())
```
